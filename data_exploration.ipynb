```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:YOUR_PASSWORD@localhost:5433/Olist"
)

print("Connection created")
```

```python
with engine.connect() as conn:
    print("PostgreSQL connection successful!")
```

```python
table_names = {
    "orders": "orders_clean_python",
    "order_items": "order_items_clean_python",
    "customers": "customers_clean_python",
    "products": "products_clean_python",
    "sellers": "sellers_clean_python",
    "reviews": "reviews_clean_python",
    "payments": "payments_clean_python",
    "category_translation": "category_translation_clean_python",
    "geolocation": "geolocation_clean_python"
}

for dataframe_name , sql_table_name in table_names.items():
    data[dataframe_name].to_sql(
        sql_table_name,
        engine,
        if_exists = "replace",
        index = False
    )

print(f"{sql_table_name} loaded : {data[dataframe_name].shape}")
```
